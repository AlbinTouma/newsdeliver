```python
import queue
import threading
```

```python
q = queue.Queue(maxsize=10)
```

```python
for _ in range(20):
    print(q.qsize())
    q.put(1)
    print(q.full())
```

```python
while not q.empty():
        item = q.get()
        print(f"Got item: {item}, Remaining: {q.qsize()}")

        q.task_done()

q.join()
```
